# Ejemplo de Sistemas de Ecuaciones

## Planteamiento:

Asumimos que tenemos observaciones de corte transversal independientes e idénticamente distribuidas:
\begin{equation*}
    \{ \mathbf{X}_i , \mathbf{Y}_i ; i = 1, 2, \ldots, N \}
\end{equation*}

Donde $\mathbf{X}_i$ es una matriz $G \times K$ y $\mathbf{Y}_i$ es un vector $G \times 1$. Así, la matriz $\mathbf{X}_i$ contiene todas las variables explicativas que aparecen en el sistema, para cada uno de los individuos indexados $i$:
\begin{equation}
    \mathbf{Y}_i = \mathbf{X}_i \boldsymbol{\beta} + \boldsymbol{\varepsilon}_i
    \label{Sist_Eqn_G}
\end{equation}

Donde $\boldsymbol{\beta}$ es un vector $K \times 1$ y $\boldsymbol{\varepsilon}_i$ es un vector $G \times 1$. De esta forma en general tendríamos:
\begin{equation*}
    \mathbf{Y}_i = 
    \begin{bmatrix}
    y_{i1} \\
    y_{i2} \\
    \vdots \\
    y_{iG}
    \end{bmatrix}
\end{equation*}

\begin{equation*}
    \boldsymbol{\varepsilon}_i = 
    \begin{bmatrix}
    \varepsilon_{i1} \\
    \varepsilon_{i2} \\
    \vdots \\
    \varepsilon_{iG}
    \end{bmatrix}
\end{equation*}

Donde $i = 1, 2, \ldots, N$. 
\begin{equation*}
    \boldsymbol{\beta} = 
    \begin{bmatrix}
    \boldsymbol{\beta}_{1} \\
    \boldsymbol{\beta}_{2} \\
    \vdots \\
    \boldsymbol{\beta}_{G}
    \end{bmatrix}
\end{equation*}

Para el caso particular de un modelo de ecuaciones aparentemente no relacionadas que no compartan variables explicativas:
\begin{equation*}
    \mathbf{x}_{ig} = [x_{i1}, x_{i2}, \ldots, x_{iK_g}]
\end{equation*}

Así, podemos proponer que la matriz $\mathbf{X}_i$ con dimensión $G \times (K_1 + K_2 + \ldots + K_G)$ es de la forma:
\begin{equation*}
    \mathbf{X}_i = 
    \begin{bmatrix}
    \mathbf{x}_{i1} \\
    \mathbf{x}_{i2} \\
    \vdots \\
    \mathbf{x}_{iG}
    \end{bmatrix} \\
\end{equation*}


## Estimadores:

El estimador de Mínimos Cuadrados Ordinarios estará dado por:
\begin{eqnarray*}
    \hat{\boldsymbol{\beta}}_{OLS} & = & (\mathbf{X}' \mathbf{X})^{-1} \mathbf{X}' \mathbf{Y}
\end{eqnarray*}

El estimador de Mínimos Cuadrados Generalizados estará dado por:
\begin{eqnarray*}
    \hat{\boldsymbol{\beta}}_{GLS} & = & (\mathbf{X}' \Sigma^{-1} \mathbf{X})^{-1} \mathbf{X}' \Sigma^{-1} \mathbf{Y}
\end{eqnarray*}

El estimador de Mínimos Cuadrados Generalizados Factibles al siguiente:
\begin{equation*}
    \hat{\hat{\boldsymbol{\beta}}} = (\mathbf{X}' \hat{\Sigma}^{-1} \mathbf{X})^{-1} \mathbf{X}' \hat{\Sigma}^{-1} \mathbf{Y}
\end{equation*}

El estimador GMM:
\begin{equation}
    \hat{\boldsymbol{\beta}} = \left( \mathbf{X}' \mathbf{Z} \hat{\mathbf{W}} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \hat{\mathbf{W}} \mathbf{Z}' \mathbf{Y} \right)
    \label{GMM_System_Beta}
\end{equation}

El estimador 2SLS:
\begin{eqnarray}
    \hat{\boldsymbol{\beta}} & = & \left( \mathbf{X}' \mathbf{Z} \left( \frac{ \mathbf{Z}'\mathbf{Z} }{N} \right)^{-1} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \left( \frac{ \mathbf{Z}'\mathbf{Z} }{N} \right)^{-1} \mathbf{Z}' \mathbf{Y} \right) \nonumber \\ 
    & = & \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}'\mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}'\mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{Y} \right) \nonumber \\
    & = & \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}'\mathbf{Z} \right)^{-1} \left( \mathbf{Z}'\mathbf{Z} \right) \left( \mathbf{Z}'\mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}'\mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{Y} \right) \nonumber \\
    & = & \left( \hat{\mathbf{X}}' \hat{\mathbf{X}} \right)^{-1} \hat{\mathbf{X}}' \mathbf{Y} 
    \label{GMM_System_Beta_2SLS}
\end{eqnarray}

El estimador 3SLS:
\begin{eqnarray}
    \hat{\boldsymbol{\beta}} & = & \left( \mathbf{X}' \mathbf{Z} \left( \frac{ \mathbf{Z}' ( \mathbf{I}_N \otimes \hat{\boldsymbol{\Sigma}} ) \mathbf{Z} }{N} \right)^{-1} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \left( \frac{ \mathbf{Z}' ( \mathbf{I}_N \otimes \hat{\boldsymbol{\Sigma}} ) \mathbf{Z} }{N} \right)^{-1} \mathbf{Z}' \mathbf{Y} \right) \nonumber \\ 
    & = & \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}' ( \mathbf{I}_N \otimes \hat{\boldsymbol{\Sigma}} ) \mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \left( \mathbf{Z}' ( \mathbf{I}_N \otimes \hat{\boldsymbol{\Sigma}} ) \mathbf{Z} \right)^{-1} \mathbf{Z}' \mathbf{Y} \right) 
    \label{GMM_System_Beta_3SLS}
\end{eqnarray}

Este ejemplo demuestra cómo se puede estimar conjuntamente un sistema de ecuaciones simultáneas utilizando mínimos cuadrados de tres etapas (3SLS). 

Las ecuaciones simultáneas modelan el salario y el número de horas trabajadas. Las dos ecuaciones son

\begin{equation}
    hours = \beta_0 + \beta_1 ln(wage) + \beta_2 educ + \beta_3 age + \beta_4 kidslt6 + \beta_5 nwifeinc + \varepsilon_i^h
\end{equation}

\begin{equation}
    ln(wage) = \gamma_0 + \gamma_1 hours + \gamma_2 educ + \gamma_3 exper + \gamma_4 exper^2 + \varepsilon_i^w
\end{equation}

Este ejemplo está basado en la discusión que hace Wooldridge (2002).

```python
# Show data

data = mroz.load()
data
```

```python
# Selecting Data and Dropping NAs values:

data = data[ ["hours", "educ", "age", "kidslt6", "nwifeinc", "lwage", "exper", "expersq"] ]

data = data.dropna()

data
```

Estos ejemplos utilizan la interfaz de fórmula. 

Esto suele ser más sencillo cuando los modelos tienen regresores exógenos, regresores endógenos e instrumentos.

Acá algunos ejemplos con el procedimiento de 2SLS:

> **Cuidado con la constante.** A diferencia de `statsmodels.formula`, las fórmulas de `linearmodels` **no agregan intercepto** por sí solas: hay que escribir `1 +` explícitamente. Sin él, el código corre sin ningún aviso, pero estima otro modelo — la recta queda forzada a pasar por el origen — y los coeficientes cambian por completo (en este ejemplo, el de `age` en la ecuación de horas pasa de −7.8 a +19.4).

```python
# Estimating hours equation:

hours = "hours ~ 1 + educ + age + kidslt6 + nwifeinc + [lwage ~ exper + expersq]" # Equation

hours_res = IV2SLS.from_formula(hours, data).fit(cov_type = "unadjusted")
# Valor en cov_type
# "unadjusted" --> Errores estándar clásicos (asumen homocedasticidad).
# "robust" --> Errores estándar robustos a heterocedasticidad (White's (1980))
# "clustered" --> Errores agrupados (útil si tienes datos por estados, familias, etc.). 
# "kernel" --> Para corregir por autocorrelación (HAC/Newey-West).
```

```python
# First Stage
print(hours_res.first_stage)
```

```python
# Second Stage
print(hours_res)
```

```python
# Estimating wage equation:

lwage = "lwage ~ 1 + educ + exper + expersq + [hours ~ age + kidslt6 + nwifeinc]"

lwage_res = IV2SLS.from_formula(lwage, data).fit(cov_type = "unadjusted")
```

```python
# First Stage
print(lwage_res.first_stage)
```

```python
# Second Stage
print(lwage_res)
```

### Verificación contra un resultado publicado

Con la constante incluida, las dos ecuaciones reproducen los ejemplos 16.5 y 16.6 de Wooldridge, *Introductory Econometrics*: en la ecuación de horas, el coeficiente de `lwage` es 1639.56 y el de `educ` es −183.75; en la de salarios, el de `hours` es 0.000126 y el de `educ` es 0.1103. Los errores estándar difieren ligeramente de los del libro porque `linearmodels` no aplica, por omisión, la corrección por grados de libertad (`debiased = False`).

Aquí el sistema se estima con `method = "ols"`, que en `IV3SLS` **no** significa MCO: es **2SLS ecuación por ecuación**, pues cada ecuación sigue usando sus propios instrumentos. Por eso las estimaciones son idénticas a las de las ecuaciones estimadas por separado.

```python
# Estimating System:

equations = dict(hours = hours, lwage = lwage)

system_2sls_res = IV3SLS.from_formula(equations, data).fit(method="ols", 
                                                           cov_type = "unadjusted")
```

```python
# 
print(system_2sls_res)
```

Con `method = "gls"` se obtiene el estimador de **3SLS** del capítulo 4 de las notas: a partir de los residuales de 2SLS se estima la matriz $\hat{\boldsymbol{\Sigma}}$ de covarianzas entre las ecuaciones, y con ella se reponderan los momentos. **Cambian las dos ecuaciones**, porque la ganancia proviene de explotar la correlación entre sus errores; si $\hat{\boldsymbol{\Sigma}}$ fuera diagonal, 3SLS coincidiría con 2SLS.

```python
# Estimating System:

equations = dict(hours = hours, lwage = lwage)

system_3sls_res = IV3SLS.from_formula(equations, data).fit(method = "gls", 
                                                           cov_type = "unadjusted")
```

### Especificación directa del modelo

El modelo se puede especificar directamente usando un diccionario de diccionarios donde los diccionarios internos contienen los 4 componentes del modelo:

* dependent - La variable dependiente
* exog - regresores exógenos
* endog - regresores endógenos
* instruments - Variables instrumentales

Las estimaciones son las mismas. Esta interfaz es más útil para generar y estimar modelos mediante programación.

```python
# La interfaz de diccionario tampoco agrega la constante: hay que incluirla como columna
data = data.assign(const = 1.0)

hours = {
    "dependent": data[["hours"]],
    "exog": data[["const", "educ", "age", "kidslt6", "nwifeinc"]],
    "endog": data[["lwage"]],
    "instruments": data[["exper", "expersq"]],
}

lwage = {
    "dependent": data[["lwage"]],
    "exog": data[["const", "educ", "exper", "expersq"]],
    "endog": data[["hours"]],
    "instruments": data[["age", "kidslt6", "nwifeinc"]],
}

equations = dict(hours = hours, lwage = lwage)

system_3sls_res = IV3SLS(equations).fit(cov_type="unadjusted")

print(system_3sls_res)
```

## Identificación de cada ecuación

Antes de comparar estimadores conviene verificar que cada ecuación esté identificada (caps. 3 y 4 de las notas):

* **Condición de orden:** al menos tantos instrumentos excluidos de la ecuación como regresores endógenos.
* **Condición de rango:** que esos instrumentos efectivamente muevan al regresor endógeno. En la práctica se juzga con la $F$ de los instrumentos excluidos en la primera etapa; la regla de Staiger y Stock pide $F > 10$.

| Ecuación | Endógena | Instrumentos excluidos | Grado de sobreidentificación |
|---|---|---|:--:|
| `hours` | `lwage` | `exper`, `expersq` | 1 |
| `lwage` | `hours` | `age`, `kidslt6`, `nwifeinc` | 2 |

Ambas ecuaciones cumplen la condición de orden. Notemos además que **las dos usan el mismo conjunto de siete variables exógenas** como instrumentos: cada ecuación excluye justamente las que la otra incluye. Ése es el caso clásico del 3SLS.

```python
# Condición de rango (F de instrumentos excluidos) y sobreidentificación (Sargan), ecuación por ecuación
identificacion = pd.DataFrame(index=["hours", "lwage"])

for nombre, res in [("hours", hours_res), ("lwage", lwage_res)]:
    diag = res.first_stage.diagnostics
    identificacion.loc[nombre, "F excluidos"] = diag["f.stat"].iloc[0]
    identificacion.loc[nombre, "p (F)"] = diag["f.pval"].iloc[0]
    identificacion.loc[nombre, "Sargan"] = res.sargan.stat
    identificacion.loc[nombre, "gl"] = res.sargan.df
    identificacion.loc[nombre, "p (Sargan)"] = res.sargan.pval

identificacion.round(4)
```

**Lectura.**

* En la ecuación de horas, `exper` y `expersq` dan una $F$ de 9.48, apenas debajo del umbral de 10. En la de salarios, `age`, `kidslt6` y `nwifeinc` dan una $F$ de **4.53**: son **instrumentos débiles** para `hours`. Por eso el coeficiente de `hours` en la ecuación de salarios (0.000126, con un error estándar del doble) es poco informativo, y hay que leerlo con las advertencias del cap. 3: sesgo hacia MCO e intervalos que no cubren lo que prometen.
* La prueba de Sargan no rechaza en ninguna de las dos ecuaciones (p = 0.35 y p = 0.23). Recordemos lo que *no* dice: si todos los instrumentos fueran inválidos en la misma dirección, la prueba no lo detectaría.

## ¿Por qué estimar el sistema? La correlación entre los errores

El 3SLS sólo mejora al 2SLS si los errores de las dos ecuaciones están correlacionados. Si $\boldsymbol{\Sigma}$ es diagonal, ambos estimadores coinciden, por la misma razón por la que SUR coincide con MCO (cap. 4 de las notas). El primer paso del 3SLS es justamente estimar
\begin{equation*}
    \hat{\boldsymbol{\Sigma}} = N^{-1} \sum_{i=1}^N \mathbf{e}_i \mathbf{e}'_i
\end{equation*}
con los residuales $\mathbf{e}_i$ del 2SLS ecuación por ecuación. Como referencia, calculamos también el análogo de la prueba de Breusch-Pagan de la sección de SUR, $LM = N r_{12}^2 \sim \chi^2_1$ bajo la hipótesis nula de que $\boldsymbol{\Sigma}$ es diagonal.

```python
import matplotlib.pyplot as plt
from scipy import stats

# Convenciones de las figuras de las notas: paleta apta para daltonismo, doble codificación por trazo
AZUL, NARANJA, VIOLETA, GRIS = "#2a78d6", "#eb6834", "#4a3aa7", "#6b6b6b"
plt.rcParams.update({
    "font.family": "serif", "font.serif": ["DejaVu Serif"], "font.size": 10,
    "axes.spines.top": False, "axes.spines.right": False,
    "figure.dpi": 110, "savefig.dpi": 200, "savefig.bbox": "tight",
})
```

```python
# Residuales del 2SLS ecuación por ecuación y matriz de covarianzas entre ecuaciones
residuales = pd.DataFrame({"hours": hours_res.resids, "lwage": lwage_res.resids})
N = len(residuales)

Sigma_hat = residuales.T @ residuales / N
r12 = residuales.corr().loc["hours", "lwage"]
LM = N * r12**2

print("Sigma estimada (2SLS):")
print(Sigma_hat.round(4))
print(f"\nCorrelación entre los errores de las dos ecuaciones: {r12:.4f}")
print(f"LM = N r^2 = {LM:.2f},  p = {stats.chi2.sf(LM, 1):.2e}")
```

```python
fig, ax = plt.subplots(figsize=(6, 4.2))
ax.scatter(residuales["hours"], residuales["lwage"], s=14, color=AZUL, alpha=0.55, edgecolor="none")
ax.axhline(0, color=GRIS, lw=0.8)
ax.axvline(0, color=GRIS, lw=0.8)
ax.set_xlabel("residual de la ecuación de horas")
ax.set_ylabel("residual de la ecuación de ln(salario)")
ax.set_title("Los errores de las dos ecuaciones no son independientes", loc="left", fontsize=11)
ax.text(0.97, 0.95, f"correlación = {r12:.2f}\nN = {N}", transform=ax.transAxes,
        ha="right", va="top", fontsize=9)
plt.show()
```

**Lectura.** La correlación es de **−0.90**, enorme frente a la de 0.16 del ejemplo de Grunfeld en `SUR_Grunfeld.ipynb`, y la prueba rechaza sin ambigüedad. Hay mucha información en una ecuación sobre los errores de la otra, que es exactamente lo que el 3SLS aprovecha.

Un cuidado de interpretación: los dos residuales son funciones de las **mismas dos variables endógenas** (`hours` y `lwage`), de modo que parte de la correlación es mecánica. No hay que leerla como un hecho causal sobre las mujeres de la muestra, sino como la estructura de covarianzas que el estimador debe tomar en cuenta.

## 2SLS y 3SLS a mano: un mismo estimador GMM con dos matrices de ponderadores

El cap. 4 de las notas muestra que 2SLS y 3SLS son casos particulares del estimador GMM del sistema,
\begin{equation*}
    \hat{\boldsymbol{\beta}} = \left( \mathbf{X}' \mathbf{Z} \hat{\mathbf{W}} \mathbf{Z}' \mathbf{X} \right)^{-1} \left( \mathbf{X}' \mathbf{Z} \hat{\mathbf{W}} \mathbf{Z}' \mathbf{Y} \right),
\end{equation*}
que sólo difieren en la elección de $\hat{\mathbf{W}}$:

* **2SLS:** $\hat{\mathbf{W}} = \left( \mathbf{Z}'\mathbf{Z} / N \right)^{-1}$.
* **3SLS:** $\hat{\mathbf{W}} = \left( N^{-1} \sum_{i} \mathbf{Z}'_i \hat{\boldsymbol{\Sigma}} \mathbf{Z}_i \right)^{-1}$, con $\hat{\boldsymbol{\Sigma}}$ de los residuales del 2SLS.

Para programarlo hay que construir las matrices **apiladas por individuo**. Con $G = 2$ ecuaciones, para cada mujer $i$:

* $\mathbf{Y}_i$ es $2 \times 1$: sus horas y su ln(salario).
* $\mathbf{X}_i$ es $2 \times 11$, diagonal por bloques: los 6 regresores de la ecuación de horas en el primer renglón y los 5 de la de salarios en el segundo.
* $\mathbf{Z}_i$ es $2 \times 14$, también diagonal por bloques: los 7 instrumentos de cada ecuación.

Al apilar las $N$ mujeres, $\mathbf{X}$ es $2N \times 11$ y $\mathbf{Z}$ es $2N \times 14$. Así $K = 11$ parámetros y $L = 14$ condiciones de momento.

```python
const = np.ones(N)

# Regresores e instrumentos de cada ecuación (mismo orden que linearmodels)
X1 = np.column_stack([const, data["educ"], data["age"], data["kidslt6"], data["nwifeinc"], data["lwage"]])
Z1 = np.column_stack([const, data["educ"], data["age"], data["kidslt6"], data["nwifeinc"], data["exper"], data["expersq"]])
X2 = np.column_stack([const, data["educ"], data["exper"], data["expersq"], data["hours"]])
Z2 = np.column_stack([const, data["educ"], data["exper"], data["expersq"], data["age"], data["kidslt6"], data["nwifeinc"]])

G = 2
K1, K2 = X1.shape[1], X2.shape[1]
L1, L2 = Z1.shape[1], Z2.shape[1]

# Apilamiento por individuo: renglón 2i para la ecuación de horas, 2i+1 para la de salarios
X = np.zeros((N * G, K1 + K2))
Z = np.zeros((N * G, L1 + L2))
Y = np.zeros(N * G)
X[0::2, :K1], X[1::2, K1:] = X1, X2
Z[0::2, :L1], Z[1::2, L1:] = Z1, Z2
Y[0::2], Y[1::2] = data["hours"], data["lwage"]

print("X:", X.shape, "  Z:", Z.shape, "  Y:", Y.shape)


def gmm_sistema(W):
    # Estimador GMM del sistema para una matriz de ponderadores W dada
    A = X.T @ Z @ W @ Z.T @ X
    return np.linalg.solve(A, X.T @ Z @ W @ Z.T @ Y)
```

```python
# 2SLS: W = (Z'Z/N)^(-1)
W_2sls = np.linalg.inv(Z.T @ Z / N)
b_2sls = gmm_sistema(W_2sls)

# Primer paso del 3SLS: Sigma con los residuales del 2SLS
E = (Y - X @ b_2sls).reshape(N, G)
Sigma = E.T @ E / N

# 3SLS: W = (sum_i Z_i' Sigma Z_i / N)^(-1)
ZSZ = sum(Z[G*i:G*i+G].T @ Sigma @ Z[G*i:G*i+G] for i in range(N))
W_3sls = np.linalg.inv(ZSZ / N)
b_3sls = gmm_sistema(W_3sls)

a_mano = pd.DataFrame({"2SLS a mano": b_2sls, "2SLS linearmodels": system_2sls_res.params.values,
                       "3SLS a mano": b_3sls, "3SLS linearmodels": system_3sls_res.params.values},
                      index=system_2sls_res.params.index)
print(a_mano.round(6))
print(f"\nDiferencia máxima, 2SLS: {np.abs(b_2sls - system_2sls_res.params.values).max():.1e}")
print(f"Diferencia máxima, 3SLS: {np.abs(b_3sls - system_3sls_res.params.values).max():.1e}")
```

Las dos fórmulas reproducen a `linearmodels` hasta el error de redondeo (en el caso del 3SLS, porque las dos ecuaciones comparten instrumentos; véase la precisión de la sección siguiente). Lo importante es el mensaje conceptual: **los dos estimadores resuelven el mismo problema de minimización**; lo único que cambia es cuánto pesa cada condición de momento. El 2SLS las trata como si los errores de las dos ecuaciones fueran independientes; el 3SLS usa $\hat{\boldsymbol{\Sigma}}$ para dar menos peso a la información redundante entre ecuaciones.

## GMM eficiente y prueba de sobreidentificación del sistema

Las notas señalan cuándo es óptima cada elección de $\hat{\mathbf{W}}$:

* el 2SLS, si $\mathbb{E}[\boldsymbol{\varepsilon}_i \boldsymbol{\varepsilon}'_i \mid \mathbf{Z}_i] = \sigma^2 \mathbf{I}_G$;
* el 3SLS, si $\mathbb{E}[\boldsymbol{\varepsilon}_i \boldsymbol{\varepsilon}'_i \mid \mathbf{Z}_i] = \boldsymbol{\Sigma}$, es decir, con correlación entre ecuaciones pero sin heterocedasticidad;
* si ninguno de los dos supuestos es defendible, la matriz óptima es $\hat{\boldsymbol{\Lambda}}^{-1}$, con $\hat{\boldsymbol{\Lambda}} = N^{-1} \sum_i \mathbf{Z}'_i \hat{\boldsymbol{\varepsilon}}_i \hat{\boldsymbol{\varepsilon}}'_i \mathbf{Z}_i$: el **GMM en dos etapas**.

`IVSystemGMM` con `weight_type = "robust"` implementa el último. Además reporta la estadística $J$ de Hansen, que contrasta las $L - K = 14 - 11 = 3$ condiciones de momento sobrantes del **sistema completo** (1 de la ecuación de horas y 2 de la de salarios).

```python
ec_hours = "hours ~ 1 + educ + age + kidslt6 + nwifeinc + [lwage ~ exper + expersq]"
ec_lwage = "lwage ~ 1 + educ + exper + expersq + [hours ~ age + kidslt6 + nwifeinc]"
sistema = dict(hours = ec_hours, lwage = ec_lwage)

system_gmm_res = IVSystemGMM.from_formula(sistema, data, weight_type = "robust").fit(cov_type = "robust")
print(system_gmm_res)
```

```python
print(system_gmm_res.j_stat)

# Verificación: GMM con la matriz de ponderadores 'unadjusted' es exactamente el 3SLS
gmm_3sls = IVSystemGMM.from_formula(sistema, data, weight_type = "unadjusted").fit(cov_type = "unadjusted")
print(f"\nGMM con W 'unadjusted' frente a 3SLS, diferencia máxima: "
      f"{np.abs(gmm_3sls.params - system_3sls_res.params).max():.1e}")
```

**Lectura.** $J = 5.82$ con 3 grados de libertad (p = 0.12): no se rechaza la validez conjunta de las condiciones de momento al 5 %. Es consistente con las dos pruebas de Sargan por ecuación, y tiene la misma limitación: sólo puede detectar instrumentos inválidos que discrepan entre sí.

La segunda línea confirma la afirmación de las notas: el 3SLS **es** un GMM con una matriz de ponderadores particular, la que supone $\mathbb{E}[\boldsymbol{\varepsilon}_i \boldsymbol{\varepsilon}'_i \mid \mathbf{Z}_i] = \boldsymbol{\Sigma}$.

> **Una precisión sobre `IV3SLS`.** La coincidencia exacta ocurre aquí porque las dos ecuaciones usan **el mismo conjunto de instrumentos**. `IV3SLS` implementa el 3SLS clásico: MCG sobre los regresores proyectados, $\left( \hat{\mathbf{X}}' (\hat{\boldsymbol{\Sigma}}^{-1} \otimes \mathbf{I}_N) \hat{\mathbf{X}} \right)^{-1} \hat{\mathbf{X}}' (\hat{\boldsymbol{\Sigma}}^{-1} \otimes \mathbf{I}_N) \mathbf{Y}$, que es un estimador del tipo GIV que mencionan las notas. Cuando cada ecuación tiene sus propios instrumentos, ese estimador **deja de coincidir** con el GMM de las notas, mientras que `IVSystemGMM` con `weight_type = "unadjusted"` sigue siendo el GMM. El ejercicio 3 lo muestra.

## La disyuntiva del 3SLS: eficiencia frente a contaminación (simulación)

Simulamos un sistema de dos ecuaciones con un regresor endógeno cada una y **coeficiente verdadero igual a 1 en ambas**:
\begin{eqnarray*}
    y_1 & = & 1 + w_1 + \varepsilon_1, \qquad w_1 = 0.5 z_1 + 0.5 z_2 + 0.5 z_4 + v_1 \\
    y_2 & = & 1 + w_2 + \varepsilon_2 + \delta z_4, \qquad w_2 = 0.5 z_3 + 0.5 z_4 + v_2
\end{eqnarray*}

* $\varepsilon_1$ y $\varepsilon_2$ tienen correlación $\rho = 0.8$: el escenario favorable al 3SLS.
* $w_g$ es endógena, porque $v_g$ está correlacionada con $\varepsilon_g$ (0.4).
* Ambas ecuaciones usan los mismos instrumentos, $(1, z_1, z_2, z_3, z_4)$, como en el ejemplo de Mroz.
* Con $\delta = 0$ todo está bien especificado. Con $\delta > 0$, la **ecuación 2** omite a $z_4$, que la afecta de forma directa. Así $z_4$ deja de ser un instrumento válido **para la ecuación 2**, mientras que la ecuación 1 sigue estando bien especificada.

La pregunta es qué le pasa al estimador de la **ecuación 1**, que no tiene ningún error.

```python
def estimar_sistema(X1, Z1, y1, X2, Z2, y2):
    # Regresa (2SLS ec. 1, 2SLS ec. 2, 3SLS ec. 1, 3SLS ec. 2), con la misma W del 3SLS que arriba
    n = len(y1)

    def tsls(Xg, Zg, yg):
        Xg_hat = Zg @ np.linalg.lstsq(Zg, Xg, rcond=None)[0]
        return np.linalg.lstsq(Xg_hat, yg, rcond=None)[0]

    b1, b2 = tsls(X1, Z1, y1), tsls(X2, Z2, y2)
    E = np.column_stack([y1 - X1 @ b1, y2 - X2 @ b2])
    S = E.T @ E / n
    # sum_i Z_i' S Z_i con Z_i diagonal por bloques
    W = np.linalg.inv(np.block([[S[0, 0] * Z1.T @ Z1, S[0, 1] * Z1.T @ Z2],
                                [S[1, 0] * Z2.T @ Z1, S[1, 1] * Z2.T @ Z2]]))
    ZX = np.block([[Z1.T @ X1, np.zeros((Z1.shape[1], X2.shape[1]))],
                   [np.zeros((Z2.shape[1], X1.shape[1])), Z2.T @ X2]])
    Zy = np.concatenate([Z1.T @ y1, Z2.T @ y2])
    b = np.linalg.solve(ZX.T @ W @ ZX, ZX.T @ W @ Zy)
    k1 = X1.shape[1]
    return b1[1], b2[1], b[1], b[k1 + 1]


def simular(delta, n=500, rho=0.8, R=2000, semilla=2026):
    rng = np.random.default_rng(semilla)
    # Correlaciones de (e1, e2, v1, v2)
    C = np.array([[1, rho, 0.4, 0], [rho, 1, 0, 0.4], [0.4, 0, 1, 0], [0, 0.4, 0, 1]])
    L = np.linalg.cholesky(C)
    uno = np.ones(n)
    salida = np.empty((R, 4))
    for r in range(R):
        z = rng.standard_normal((n, 4))
        e1, e2, v1, v2 = (rng.standard_normal((n, 4)) @ L.T).T
        w1 = 0.5 * z[:, 0] + 0.5 * z[:, 1] + 0.5 * z[:, 3] + v1
        w2 = 0.5 * z[:, 2] + 0.5 * z[:, 3] + v2
        y1 = 1 + w1 + e1
        y2 = 1 + w2 + e2 + delta * z[:, 3]
        Zc = np.column_stack([uno, z])
        salida[r] = estimar_sistema(np.column_stack([uno, w1]), Zc, y1,
                                    np.column_stack([uno, w2]), Zc, y2)
    return pd.DataFrame(salida, columns=["2SLS ec. 1", "2SLS ec. 2", "3SLS ec. 1", "3SLS ec. 2"])


sim_bien = simular(delta=0.0)
sim_mal = simular(delta=0.3)

resumen = pd.concat({"delta = 0 (bien especificado)": sim_bien.agg(["mean", "std"]),
                     "delta = 0.3 (ec. 2 mal especificada)": sim_mal.agg(["mean", "std"])})
resumen.round(3)
```

```python
# Error cuadrático medio de la ecuación 1 según la magnitud del error de especificación
deltas = np.linspace(0, 0.5, 11)
ecm = []
for d in deltas:
    s = simular(delta=d, R=1000)
    ecm.append({"delta": d,
                "2SLS": np.sqrt(((s["2SLS ec. 1"] - 1) ** 2).mean()),
                "3SLS": np.sqrt(((s["3SLS ec. 1"] - 1) ** 2).mean())})
ecm = pd.DataFrame(ecm).set_index("delta")
ecm.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
bins = np.linspace(0.7, 1.25, 38)

for ax, sim, titulo in [(axes[0], sim_bien, "(a) Bien especificado: δ = 0"),
                        (axes[1], sim_mal, "(b) Ecuación 2 mal especificada: δ = 0.3")]:
    ax.hist(sim["2SLS ec. 1"], bins=bins, histtype="step", lw=2, color=AZUL, density=True)
    ax.hist(sim["3SLS ec. 1"], bins=bins, histtype="step", lw=2, color=NARANJA, ls="--", density=True)
    ax.axvline(1, color=GRIS, lw=1, ls=":")
    ax.set_xlabel("estimación del coeficiente de la ecuación 1")
    ax.set_title(titulo, loc="left", fontsize=11)
    ax.set_yticks([])
    ax.spines["left"].set_visible(False)
from matplotlib.lines import Line2D
axes[0].legend(handles=[Line2D([], [], color=AZUL, lw=2, label="2SLS"),
                        Line2D([], [], color=NARANJA, lw=2, ls="--", label="3SLS"),
                        Line2D([], [], color=GRIS, lw=1, ls=":", label="valor verdadero")],
               loc="upper left", frameon=False, fontsize=9)

# (c) Raíz del ECM contra la magnitud del error de especificación
ax = axes[2]
ax.plot(ecm.index, ecm["2SLS"], color=AZUL, lw=2, marker="o", ms=5)
ax.plot(ecm.index, ecm["3SLS"], color=NARANJA, lw=2, ls="--", marker="s", ms=5)
ax.text(0.5, ecm["2SLS"].iloc[-1] + 0.007, "2SLS", color=AZUL, ha="right", fontsize=9.5)
ax.text(0.3, ecm["3SLS"].iloc[6] + 0.01, "3SLS", color=NARANJA, ha="right", fontsize=9.5)
ax.set_xlabel("δ: efecto omitido de z₄ en la ecuación 2")
ax.set_ylabel("raíz del ECM, ecuación 1")
ax.set_ylim(0, None)
ax.set_title("(c) Ganancia y riesgo", loc="left", fontsize=11)

plt.tight_layout()
plt.show()
```

**Lectura.**

* **(a) Bien especificado.** Los dos estimadores están centrados en 1, pero la distribución del 3SLS es mucho más estrecha: su desviación estándar es cerca de 40 % menor. Ésa es la ganancia de eficiencia que promete la teoría cuando $\rho$ es alto.
* **(b) La ecuación 2 omite una variable.** El 2SLS de la **ecuación 1** sigue centrado en 1, porque cada ecuación se estima por separado y el error queda confinado a la ecuación 2. El 3SLS de la ecuación 1, en cambio, se **desplaza a cerca de 0.90**, aunque esa ecuación no tiene ningún error. El mecanismo es la matriz $\hat{\mathbf{W}}$: al ponderar conjuntamente las condiciones de momento de las dos ecuaciones, la condición violada de la ecuación 2 arrastra la estimación de la ecuación 1.
* **(c) La disyuntiva.** Para errores de especificación pequeños, el 3SLS sigue teniendo menor error cuadrático medio: la reducción de varianza compensa el sesgo. Las curvas se cruzan entre $\delta = 0.10$ y $\delta = 0.15$; a partir de ahí el sesgo domina y el 2SLS es preferible. Notemos que $\delta = 0.15$ es un error modesto, y que en datos reales nadie conoce $\delta$.

Por eso, como dicen las notas, en trabajo aplicado conviene **reportar ambos**: si 2SLS y 3SLS difieren mucho más de lo que justifica la diferencia de precisión, es señal de que alguna ecuación está mal especificada. El ejercicio 6 formaliza esa comparación como una prueba de Hausman.

## Conclusiones

* Las dos ecuaciones del modelo de Mroz cumplen la condición de orden, pero la de salarios tiene **instrumentos débiles** ($F = 4.53$).
* Los errores de las dos ecuaciones tienen una correlación de −0.90, de modo que estimar el sistema conjuntamente tiene un potencial grande de ganancia.
* 2SLS, 3SLS y GMM eficiente son **un mismo estimador GMM** con tres matrices de ponderadores distintas; programados a mano, coinciden con `linearmodels`.
* El 3SLS reduce los once errores estándar; la prueba $J$ no rechaza las restricciones de sobreidentificación del sistema.
* La ganancia tiene un precio: un error en una ecuación contamina a las demás. Reportar 2SLS y 3SLS juntos es la forma práctica de vigilarlo.
