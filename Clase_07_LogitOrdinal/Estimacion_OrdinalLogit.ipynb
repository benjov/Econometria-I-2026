### Estimación (Running the ordered logit model)

```python
# Define dependent and regresors variables
X = datos_subset[[ 'Price', 'Score', 'Followers_n', 'positive_n', 'Adventure', 'Singleplayer', 'Story Rich',
                   'Horror', '3D', 'Survival', 'Choices Matter', 'Historical', 'Mystery', 'Replay Value',
                   'Early Access', 'Female Protagonist', 'Lovecraftian', 'Open World', 'Multiplayer', 'Co-op',
                   'Indie', 'Atmospheric', 'Violent', 'Online Co-Op', 'Third Person', 'Sandbox', 'Strategy',
                   'Difficult', 'Gore', 'Soundtrack', 'Great Soundtrack', 'Simulation', 'RPG',
                   'Character Customization', 'Funny', 'Sci-fi', 'Exploration', 'Action', 'Sexual Content' ]]

y = datos_subset['Cat.Rev.P']
```

```python
# https://www.statsmodels.org/stable/generated/statsmodels.miscmodels.ordinal_model.OrderedModel.html
modelo = OrderedModel(y, X, distr = 'logit')

resultados = modelo.fit(method = 'bfgs')

print(resultados.summary())
```

### Efectos Marginales del Logit Ordinal

El objeto `OrderedResults` de statsmodels **no** tiene el método `get_margeff()`. Los efectos marginales se calculan manualmente. Para cada variable $x_k$ y cada categoría $j$, el efecto marginal es:

$$\frac{\partial P(Y=j)}{\partial x_k} = [\Lambda'(\tau_{j-1} - X\beta) - \Lambda'(\tau_j - X\beta)] \cdot \beta_k$$

donde $\Lambda'(z) = \Lambda(z)(1 - \Lambda(z))$ es la densidad logística. Se implementan dos versiones:
- **MEM** (Marginal Effects at the Mean): se evalúa en el vector de medias muestral.
- **AME** (Average Marginal Effects): se promedia el efecto sobre toda la muestra.

```python
# ============================================================
# Efectos Marginales en la Media (MEM)
# ============================================================
import numpy as np
import pandas as pd

# --- Extraer parametros ---
params    = resultados.params
n_vars    = len(X.columns)
n_cats    = len(y.cat.categories)
betas     = params[:n_vars].values
thresholds = params[n_vars:].values

# Reconstruir umbrales: el primero es libre, los siguientes son log(delta)
cuts = np.zeros(n_cats - 1)
cuts[0] = thresholds[0]
for i in range(1, n_cats - 1):
    cuts[i] = cuts[i-1] + np.exp(thresholds[i])

# Punto de evaluacion: media muestral
X_num = X.copy()
for col in X_num.select_dtypes(['category']).columns:
    X_num[col] = X_num[col].astype(float)
X_mean = X_num.mean().values
Xb_mean = X_mean @ betas

# Funcion logistica y su derivada
def logistic(z):     return 1 / (1 + np.exp(-z))
def logistic_pdf(z): lam = logistic(z); return lam * (1 - lam)

extended_cuts = np.concatenate([[-np.inf], cuts, [np.inf]])
pdfs_mean = np.array([logistic_pdf(c - Xb_mean) for c in extended_cuts])

cat_labels = y.cat.categories.tolist()
var_labels  = X.columns.tolist()

me_matrix = np.zeros((n_vars, n_cats))
for k in range(n_vars):
    for j in range(n_cats):
        me_matrix[k, j] = (pdfs_mean[j] - pdfs_mean[j+1]) * betas[k]

df_mem = pd.DataFrame(me_matrix, index=var_labels, columns=cat_labels)

print('Efectos Marginales en la Media (MEM) — dP(Y=j)/dx_k')
print('=' * 80)
print(df_mem.round(6).to_string())
```

```python
# ============================================================
# Efectos Marginales Promedio (Average Marginal Effects, AME)
# ============================================================
# Se calcula el efecto marginal en cada observacion y se promedia.
# Recomendado para variables binarias/categoricas.

X_arr = X_num.values   # matriz (N x K)
N     = X_arr.shape[0]

ame_matrix = np.zeros((n_vars, n_cats))

for i in range(N):
    Xb_i   = X_arr[i] @ betas
    pdfs_i = np.array([logistic_pdf(c - Xb_i) for c in extended_cuts])
    for k in range(n_vars):
        for j in range(n_cats):
            ame_matrix[k, j] += (pdfs_i[j] - pdfs_i[j+1]) * betas[k]

ame_matrix /= N

df_ame = pd.DataFrame(ame_matrix, index=var_labels, columns=cat_labels)

print('Efectos Marginales Promedio (AME) — dP(Y=j)/dx_k')
print('=' * 80)
print(df_ame.round(6).to_string())
```

```python
# ============================================================
# Visualizacion: AME sobre P(Y = '4. Mas de 20M')
# ============================================================
import matplotlib.pyplot as plt

target_cat = '4. Mas de 20M'
# Usar la ultima columna si el nombre exacto no coincide
col_target = df_ame.columns[-1]

ame_top = (
    df_ame[col_target]
    .reindex(df_ame[col_target].abs().sort_values(ascending=False).index)
    .head(20)
)

colors = ['#2ecc71' if v > 0 else '#e74c3c' for v in ame_top]

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(ame_top.index[::-1], ame_top.values[::-1], color=colors[::-1])
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Efecto Marginal Promedio (AME)')
ax.set_title(f"Top 20 AME sobre P(Y = '{col_target}')")
ax.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Heatmap de AME — solo variables significativas (p < 0.05)
# ============================================================
import seaborn as sns

pvals    = resultados.pvalues[:n_vars]
sig_vars = pvals[pvals < 0.05].index.tolist()

df_ame_sig = df_ame.loc[sig_vars]

fig, ax = plt.subplots(figsize=(10, max(4, len(sig_vars) * 0.5)))
sns.heatmap(
    df_ame_sig,
    annot=True, fmt='.4f', cmap='RdYlGn', center=0,
    linewidths=0.5, ax=ax, cbar_kws={'label': 'AME'}
)
ax.set_title('Efectos Marginales Promedio (AME) — Variables Significativas (p<0.05)')
ax.set_xlabel('Categoria de Ingresos')
ax.set_ylabel('Variable')
plt.tight_layout()
plt.show()
```
